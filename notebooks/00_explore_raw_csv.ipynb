```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/entsoe/load_RO_2026-09-13_2026-09-20.csv")
df_copy = df.copy()

print(df_copy.head())
```

```python
print(df_copy.shape)
```

```python
print(df_copy.dtypes)
```

```python

df_copy["Actual Total Load (MW)"] = pd.to_numeric(df_copy["Actual Total Load (MW)"], errors="coerce")
df_copy["Day-ahead Total Load Forecast (MW)"] = pd.to_numeric(df_copy["Day-ahead Total Load Forecast (MW)"], errors="coerce")

print(df_copy["Actual Total Load (MW)"].isna().sum())

mask = df_copy["Actual Total Load (MW)"].isna()
print(df["Actual Total Load (MW)"][mask].unique())

print(df_copy[mask][["MTU (UTC)"]])
```

```python
df_copy["mtu_start"] = pd.to_datetime(df_copy["MTU (UTC)"].str.split(" - ").str[0], format='%d/%m/%Y %H:%M')
df_copy["mtu_start"] = df_copy["mtu_start"].dt.tz_localize("UTC")
df_hourly = df_copy[["mtu_start", "Actual Total Load (MW)"]].resample("1h", on="mtu_start").mean()

print(df_hourly.head())
```
